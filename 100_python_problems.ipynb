````python
import IPython.display as dis

def inputdefaultnumber[T: (int, float)](prompt: str, typename: type[T] = int) -> T | None:
  strnum = input(prompt)

  try:
    num = typename(strnum)
  except:
    num = None
    dis.clear_output()

  return num

def inputnumber[T: (int, float)](prompt: str, typename: type[T] = int) -> T:
  num = inputdefaultnumber(prompt, typename)

  while num is None:
    num = inputdefaultnumber(prompt, typename)

  return num
````

````python
import IPython.display as dis

def input_numbers[T: (int, float)](*args: str, typename: type[T] = int) -> list[T]:
  nums: list[T] = []

  index = 0
  while index < len(args):
    prompt = args[index]
    strnum = input(prompt).strip()

    dis.clear_output()

    try:
      nums.append(typename(strnum))
      print(nums)
    except:
      continue

    index += 1

  return nums
````

## 8 Euclidean distance
write aprogram to find the euclidean distance between two coordinates

````python
import IPython.display as dis

coords = []
while len(coords) < 4:
  index = len(coords) // 2
  strx = input(f"enter x{index} coordinate: ")
  stry = input(f"enter y{index} coordinate: ")

  if strx.isnumeric() and stry.isnumeric():
    coords.extend([int(strx), int(stry)])
    print(f"[x{index}, y{index}] =", coords[index*2: index*2+2])
  else:
    dis.clear_output()

ed = pow((coords[0] - coords[2])**2 + (coords[1] - coords[3])**2, 1/2)
print(f"euclidean distance = {ed:.2f}")
````

## 10 Profit or loss calculator
Write a program that will take user input of cost price and selling price and determine whether it's a loss or profit


````python
import IPython.display as dis

def isfloat(strnum: str):
  try:
    float(strnum)
    return True
  except:
    return False

while True:
  strbuying_price = input("enter buying price $: ")
  strselling_price = input("enter selling price $: ")

  if isfloat(strbuying_price) and isfloat(strselling_price):
    buying_price = float(strbuying_price)
    selling_price = float(strselling_price)

    diff = selling_price - buying_price
    break

diffp = (diff * 100) / buying_price
print("profit ✅" if diff > 0 else "loss ❌", f"{diff}$ {diffp:.2f}%")
````

## 14 Hour and Minute hands angle
Given a string s representing time in 24-hour format "HH:MM", compute the smallest angle in degrees between the hour and minute hands of an analog clock.

```md
Input: s = "06:00"
Output: 180.000
Explanation: When the time is 06:00, the angle between the hour and minute hands of the clock is 180.000 degrees.
```

```md
Input: s = "03:15"
Output: 7.500
Explanation: When the time is 03:15, the angle between the hour and minute hands of the clock is 7.500 degrees.
```

````python
while True:
  hours = inputnumber("enter hours 24h: ", int)
  if hours is None or not (hours >= 0 and hours < 24):
    continue

  mins = inputnumber("enter minutes 60m: ", int)
  if mins is None or not (mins >= 0 and mins < 60):
    continue

  minsangle = mins * 6 # 90
  hoursangle = (hours % 12) * 30 + (minsangle / 12)
  break

diff = abs(hoursangle - minsangle)
idiff = 360 - diff
print(f"smallest angle {hours}:{mins} = {diff if diff < idiff else idiff} ✅")
````

* A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle

* Two rectangles sharing a side are considered overlapping. (`l1[]` and `r1[]` are the extreme points of the first rectangle `l2[]` and `r2[]` are the extreme points of the second rectangle)

*Note: It may be assumed that all sides of rectables are parallel to the coordinate axis (2 sides parallel to X axis and other sides to Y axis)*

**Examples:**
```md
Input: l1[] = [0, 10], r1[] = [10, 0], l2[] = [5, 5], r2[] = [15, 0]
Output: true
Explanation: The rectangles overlap.
```

```md
Input: l1[] = [0, 2], r1[] = [1, 1], l2[] = [-2, 0], r2[] = [0, -3]
Output: false
Explanation: The rectangles do not overlap.
```

````python
import typing

class Point(typing.NamedTuple):
  x: int
  y: int

def isin_rectagle_area(p: Point, q1: Point, q2: Point) -> bool:
  return q1.x < p.x < q2.x or q1.y < p.y < q2.y
````

## 18 Armstrong number checker
Write a program that will check whether the number is an Armstrong number or not

````python
import math

def isarmstrong_number(num: int):
  numcopy = num
  sumdigits = 0
  tdigits = math.trunc(math.log10(num)) + 1

  while num > 0:
    sumdigits += (num % 10)**tdigits
    num //= 10

  return numcopy == sumdigits
````

````python
[num] = input_numbers("enter number: ")

isan = isarmstrong_number(num)
print(f"armstrong number {"✅" if isan else "❌"} {num}")
````

## 21 Unit Converter
Write a menu-driven program

1. `cm` to `ft`
2. `km` to `miles`
3. `USD` to `INR`
4. exit


````python
def convertcm2ft(cm: int) -> float:
  return cm / 30.48

def convertkm2miles(km: int) -> float:
  return km / 1.609344

def convertusd2inr(usd: float) -> float:
  return usd * 83.825
````

````python
import ipywidgets as widgets
import google.colab as colab
import IPython.display as display

options = {
    1: ("cm to ft", convertcm2ft),
    2: ("km to miles", convertkm2miles),
    3: ("USD to INR", convertusd2inr),
    4: "exit",
}

while True:
  for key, opt in options.items():
    if key < 4:
      view, callback = opt
      print(f"{key} {view}")
    else:
      print(f"{key} {opt}")

  selectedopt = inputnumber("enter option: ")

  if not (selectedopt in options.keys()):
    display.clear_output()
    continue

  if selectedopt > 3:
    break

  view, callback = options[selectedopt]
  base, to, target = view.split(" ")

  display.clear_output()
  value = inputnumber(f"enter {base}: ")
  result = callback(value)
  print(f"{value} {base} = {result:.2f} {target} ✅\n")
````
