## 1 Find the Oldest Age:
User will input three ages. Find the oldest one.

````python
import IPython.display as dis

ages = []
while len(ages) < 3:
  strage = input(f"enter age {len(ages) + 1}:")

  if strage.isnumeric():
    ages.append(int(strage))

  dis.clear_output()
  print(ages)

ages.sort()
maxage = ages[-1]
print(f"max age is : {maxage}")
````

## 2 Celsius to Fahrenheit Converter:
Write a program that will convert Celsius value to Fahrenheit.

````python
import IPython.display as dis

while True:
  strcelsius = input("enter celsius value:")

  try:
    celsius = float(strcelsius)
    fahrenheit = (celsius * 9/5) + 32
    print(f"{celsius} celsius is {fahrenheit:.2f} fahrenheit")
    break
  except:
    dis.clear_output()
````

## 3 Swap two numbers

## 5 Reverse a number and check
Write a program that will reverse a number and check whether the reverse is the same as the original

````python
import IPython.display as dis

while True:
  strnum = input("enter number: ")

  if strnum.isnumeric():
    revnum = strnum[::-1]

    if strnum == revnum:
      print(f"{strnum} is equal to {revnum} ✅")
      break
    else:
      print(f"{strnum} is not equal to {revnum} ❌")
  else:
    dis.clear_output()
````

## 8 Euclidean distance
write aprogram to find the euclidean distance between two coordinates

````python
import IPython.display as dis

coords = []
while len(coords) < 4:
  index = len(coords) // 2
  strx = input(f"enter x{index} coordinate: ")
  stry = input(f"enter y{index} coordinate: ")

  if strx.isnumeric() and stry.isnumeric():
    coords.extend([int(strx), int(stry)])
    print(f"[x{index}, y{index}] =", coords[index*2: index*2+2])
  else:
    dis.clear_output()

ed = pow((coords[0] - coords[2])**2 + (coords[1] - coords[3])**2, 1/2)
print(f"euclidean distance = {ed:.2f}")
````

## 10 Profit or loss calculator
Write a program that will take user input of cost price and selling price and determine whether it's a loss or profit


````python
import IPython.display as dis

def isfloat(strnum: str):
  try:
    float(strnum)
    return True
  except:
    return False

while True:
  strbuying_price = input("enter buying price $: ")
  strselling_price = input("enter selling price $: ")

  if isfloat(strbuying_price) and isfloat(strselling_price):
    buying_price = float(strbuying_price)
    selling_price = float(strselling_price)

    diff = selling_price - buying_price
    break

diffp = (diff * 100) / buying_price
print("profit ✅" if diff > 0 else "loss ❌", f"{diff}$ {diffp:.2f}%")
````

## 11 Simple Interest Calculator
Write a program to find the simple interest when the value of the principal, rate of interest, and time period given

````python
import IPython.display as dis

def inputnumber[T: (int, float)](prompt: str, typename: type[T]) -> T | None:
  strnum = input(prompt)

  try:
    num = typename(strnum)
  except:
    num = None
    dis.clear_output()

  return num
````

````python
while True:
  val = inputnumber("enter value $: ", int)
  if val is None:
    continue

  irate = inputnumber("enter interest rate %: ", float)
  if irate is None or not (irate > 0 and irate < 100):
    continue

  tperiod = inputnumber("enter time period (months): ", int)
  if tperiod is None:
    continue

  total = val * (1 + irate * tperiod / 100)
  break

print(f"total = {total:.2f} ✅")
````

## 12 Volume of Cylinder and Cost Calculator

## 15 Overlapping rectangles

* A rectangle is denoted by providing the x and y coordinates of two points: the left top corner and the right bottom corner of the rectangle

* Two rectangles sharing a side are considered overlapping. (`l1[]` and `r1[]` are the extreme points of the first rectangle `l2[]` and `r2[]` are the extreme points of the second rectangle)

*Note: It may be assumed that all sides of rectables are parallel to the coordinate axis (2 sides parallel to X axis and other sides to Y axis)*

**Examples:**
```md
Input: l1[] = [0, 10], r1[] = [10, 0], l2[] = [5, 5], r2[] = [15, 0]
Output: true
Explanation: The rectangles overlap.
```

```md
Input: l1[] = [0, 2], r1[] = [1, 1], l2[] = [-2, 0], r2[] = [0, -3]

Explanation: The rectangles do not overlap.
```
