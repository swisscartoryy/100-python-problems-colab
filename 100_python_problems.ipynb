## 1. Find the Oldest Age:
User will input three ages. Find the oldest one.

```python
import IPython.display as dis

ages = []
while len(ages) < 3:
  strage = input(f"enter age {len(ages) + 1}:")

  if strage.isnumeric():
    ages.append(int(strage))

  dis.clear_output()
  print(ages)

ages.sort()
maxage = ages[-1]
print(f"max age is : {maxage}")
```

## 2. Celsius to Fahrenheit Converter:
Write a program that will convert Celsius value to Fahrenheit.

```python
import IPython.display as dis

while True:
  strcelsius = input("enter celsius value:")

  try:
    celsius = float(strcelsius)
    fahrenheit = (celsius * 9/5) + 32
    print(f"{celsius} celsius is {fahrenheit:.2f} fahrenheit")
    break
  except:
    dis.clear_output()
```
